```python
import copy

# 1. 共同設定：讀取完整學號
student_id = input("請輸入完整學號：").strip().upper()
tail = student_id[-3:]
a = int(student_id[-3])
b = int(student_id[-2])
c = int(student_id[-1])
s = a + b + c

# 2. 準備兩場電子報活動的原始資料
saved = ["BASE"]
raw_a = [" welcome" + tail + " ", "vip" + tail, "", "WELCOME" + tail, " "]
raw_b = [" new" + tail + " ", " welcome" + tail + " "]

# 3. 撰寫整理優惠碼的函式
def collect_codes(raw_codes, saved=None):
    # 若未傳入 saved，從新的空串列開始；若有提供則建立副本以避免修改原始 saved
    if saved is None:
        result = []
    else:
        result = list(saved)

    # 整理 raw_codes
    for code in raw_codes:
        clean_code = code.strip().upper()
        # 略過空字串
        if clean_code == "":
            continue
        # 相同優惠碼只保留一次，順序以首次出現為準
        if clean_code not in result:
            result.append(clean_code)

    return result

# 複製原始資料以驗證原始串列未被修改
saved_backup = list(saved)
raw_a_backup = list(raw_a)

# 4. 依序計算活動 A 與活動 B 的結果
activity_a = collect_codes(raw_a, saved=saved)
activity_b = collect_codes(raw_b)

# 5. 印出結果驗證
print(f"學號：{student_id}")
print(f"活動A：{activity_a}")
print(f"活動B：{activity_b}")
print(f"原始saved：{saved_backup}")
print(f"原始raw_a：{raw_a_backup}")

# 6. 呼叫空資料並印出結果
empty_test = collect_codes([])
print(f"空資料：{empty_test}")
```
