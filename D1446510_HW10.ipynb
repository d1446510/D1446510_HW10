# 共同設定：學號讀取與變數定義

```python
# 讀取完整學號並進行格式化處理
student_id = input("請輸入完整學號：").strip().upper()

# 提取末三碼及各別數字
tail = student_id[-3:]
a = int(student_id[-3])
b = int(student_id[-2])
c = int(student_id[-1])
s = a + b + c

# 驗證輸出結果
print(f"\n學號 (student_id): {student_id}")
print(f"末三碼 (tail): {tail}")
print(f"a = {a}, b = {b}, c = {c}")
print(f"s (a + b + c) = {s}")
```

## 第一題：製作自己的廣告成效報表

```python
def ad_report(impressions, clicks, target=3.0):
    # 曝光次數為 0 時，回傳 (None, False)
    if impressions == 0:
        return (None, False)

    # 計算 CTR 百分比
    ctr = (clicks / impressions) * 100

    # 判斷是否達標
    passed = ctr >= target
    return (ctr, passed)

# 產生符合學號變數的廣告數據
ads = [
    ("搜尋廣告", 1000 + 100 * a, 30 + 5 * b),
    ("社群廣告", 1200 + 100 * b, 24 + 4 * c),
    ("影音廣告", 0, 0),
]

# 個人 CTR 門檻
target = 2.5 + 0.1 * c

# 逐筆計算並印出報表
print(f"學號：{student_id}")
print(f"CTR門檻：{target:.2f}%")

passed_count = 0
for name, imp, clk in ads:
    ctr, passed = ad_report(imp, clk, target=target)

    # 格式化 CTR 顯示
    if ctr is None:
        ctr_display = "無曝光資料"
    else:
        ctr_display = f"{ctr:.2f}%"

    if passed:
        passed_count += 1

    print(f"{name}：曝光={imp}，點擊={clk}，CTR={ctr_display}，達標={passed}")

print(f"達標廣告數：{passed_count}")

# 驗證預設門檻與零點擊測試
default_test = ad_report(1000, 30)
zero_click_test = ad_report(1000, 0)

print(f"預設門檻測試：{default_test}")
print(f"零點擊測試：{zero_click_test}")
```

## 第二題：整理兩場電子報活動的個人優惠碼

```python
def collect_codes(raw_codes, saved=None):
    # 每次回傳新的串列，若未傳入 saved，從新的空串列開始，不修改原本傳入的 saved 串列
    result = list(saved) if saved is not None else []

    for raw_code in raw_codes:
        cleaned = raw_code.strip().upper()
        # 略過空字串
        if cleaned == "":
            continue
        # 相同優惠碼只保留一次，順序以首次出現為準
        if cleaned not in result:
            result.append(cleaned)

    return result

# 建立資料
saved = ["BASE"]
raw_a = [" welcome" + tail + " ", "vip" + tail, "", "WELCOME" + tail, "   "]
raw_b = [" new" + tail + " ", " welcome" + tail + " "]

# 複製一份原始 raw_a 和 saved 來比對是否被函式內部的運算修改
raw_a_orig = list(raw_a)
saved_orig = list(saved)

# 呼叫與計算
activity_a = collect_codes(raw_a, saved=saved)
activity_b = collect_codes(raw_b)
empty_test = collect_codes([])

# 輸出結果
print(f"學號：{student_id}")
print(f"活動A：{activity_a}")
print(f"活動B：{activity_b}")
print(f"原始saved：{saved_orig}")
print(f"原始raw_a：{raw_a_orig}")
print(f"空資料：{empty_test}")
```

## 第三題：遞迴計算個人活動的累計廣告費

```python
def cumulative_spend(daily_spend, n):
    # 遞迴停止條件：n 為 0 時回傳 0
    if n == 0:
        return 0
    # 遞迴步驟：前 n-1 天的費用加上第 n 天的費用 (索引為 n-1)
    return cumulative_spend(daily_spend, n - 1) + daily_spend[n - 1]

# 建立資料
daily_spend = [
    100 + 10 * a,
    150 + 10 * b,
    0,
    200 + 10 * c,
    50 + 10 * s
]
n = 3 + c % 3

# 輸出學號、每日費用與指定 n 天累計費用
print(f"學號：{student_id}")
print(f"每日費用：{daily_spend}")
print(f"指定n：{n}")
print(f"指定前{n}天累計：{cumulative_spend(daily_spend, n)}元")

# 依序印出前 0、1、2、3、4、5 天的累計費用
for i in range(6):
    total = cumulative_spend(daily_spend, i)
    print(f"前{i}天：{total}元")

# 空資料測試
empty_test = cumulative_spend([], 0)
print(f"空資料：{empty_test}元")
```

## 第四題：清理素材點擊資料並產生排名

```python
def select_clicks(raw_clicks, min_clicks=10):
    # 1. map() 搭配 lambda，去掉每筆文字的頭尾空白
    stripped = map(lambda x: x.strip(), raw_clicks)

    # 2. filter(None, ...)，刪掉空字串
    non_empty = filter(None, stripped)

    # 3. map(int, ...) 轉整數，並用 list() 保存結果
    numbers = list(map(int, non_empty))

    # 4. 使用 filter() 搭配 lambda 篩選大於或等於 min_clicks 的項目
    qualified = list(filter(lambda x: x >= min_clicks, numbers))

    return (numbers, qualified)

# 建立資料
raw_clicks = [
    str(10 + a),
    "0",
    "",
    " " + str(5 + b) + " ",
    "   ",
    str(15 + c),
    str(5 + b)
]
min_clicks = 5 + b

# 呼叫函式進行清理與篩選
numbers, qualified = select_clicks(raw_clicks, min_clicks=min_clicks)

# 印出基本結果
print(f"學號：{student_id}")
print(f"點擊門檻：{min_clicks}")
print(f"清理結果：{numbers}")
print(f"合格結果：{qualified}")
print(f"合格筆數：{len(qualified)}")
print(f"合格合計：{sum(qualified)}")

# 排序並產生排名
sorted_qualified = sorted(qualified, reverse=True)
for rank, clicks in enumerate(sorted_qualified, start=1):
    print(f"第{rank}名：{clicks}次")

# 使用門檻 100 進行測試
numbers_100, qualified_100 = select_clicks(raw_clicks, min_clicks=100)
print(f"門檻100合格結果：{qualified_100}")
print(f"門檻100合格筆數：{len(qualified_100)}")
print(f"門檻100合格合計：{sum(qualified_100)}")
if not qualified_100:
    print("無符合門檻的素材")
```
